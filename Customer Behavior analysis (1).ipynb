```python
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\munna\Downloads\customer_shopping_behavior (1).csv")
```

```python
# Filling Blank Rows for review rating
df['Review Rating'] = (
    df.groupby('Category')['Review Rating']
      .transform(lambda x: x.fillna(x.median()))
)
```

```python
df.isnull().sum()
```

```python
# converting columns name into snake case
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#creating age_group
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels=labels)
df[['age','age_group']].head(10)
```

```python
# Creating purchase Frequency Days
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quartely' : 90,
    'Bi-Weekly' : 14,
    'Anually' : 365,
    'Every 3 Month' : 90,
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df =df.drop('promo_code_used',axis=1)
```

```python
### Exporting lean Csv file
df.to_csv("customer_shopping_behavior_clean_data",index=False)
```
